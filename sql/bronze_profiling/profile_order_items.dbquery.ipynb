```sql
-- ====================================================================
-- ORDER ITEMS TABLE PROFILING
-- ====================================================================

-- Basic counts
SELECT 
    COUNT(*) as total_items,
    COUNT(DISTINCT order_id) as unique_orders,
    COUNT(DISTINCT product_id) as unique_products,
    COUNT(DISTINCT seller_id) as unique_sellers
FROM workspace.retail_bronze.bronze_order_items;

-- Null analysis
SELECT 
    COUNT(*) - COUNT(order_id) as null_order_id,
    COUNT(*) - COUNT(product_id) as null_product_id,
    COUNT(*) - COUNT(seller_id) as null_seller_id,
    COUNT(*) - COUNT(price) as null_price,
    COUNT(*) - COUNT(freight_value) as null_freight
FROM workspace.retail_bronze.bronze_order_items;

-- Price statistics
SELECT 
    MIN(price) as min_price,
    MAX(price) as max_price,
    AVG(price) as avg_price,
    PERCENTILE(price, 0.5) as median_price
FROM workspace.retail_bronze.bronze_order_items;

-- Freight statistics
SELECT 
    MIN(freight_value) as min_freight,
    MAX(freight_value) as max_freight,
    AVG(freight_value) as avg_freight
FROM workspace.retail_bronze.bronze_order_items;

-- Check for invalid prices (negative or zero)
SELECT 
    COUNT(*) as invalid_price_count
FROM workspace.retail_bronze.bronze_order_items
WHERE price <= 0 OR freight_value < 0;
```
