```sql
-- ====================================================================
-- ORDERS TABLE PROFILING
-- ====================================================================

-- Basic counts
SELECT 
    COUNT(*) as total_rows,
    COUNT(DISTINCT order_id) as unique_orders,
    COUNT(DISTINCT customer_id) as unique_customers
FROM workspace.retail_bronze.bronze_orders;

-- Null analysis
SELECT 
    COUNT(*) - COUNT(order_id) as null_order_id,
    COUNT(*) - COUNT(customer_id) as null_customer_id,
    COUNT(*) - COUNT(order_status) as null_status,
    COUNT(*) - COUNT(order_purchase_timestamp) as null_purchase_date,
    COUNT(*) - COUNT(order_delivered_customer_date) as null_delivered_date
FROM workspace.retail_bronze.bronze_orders;

-- Order status distribution
SELECT 
    order_status,
    COUNT(*) as count,
    ROUND(COUNT(*) * 100.0 / (SELECT COUNT(*) FROM workspace.retail_bronze.bronze_orders), 2) as percentage
FROM workspace.retail_bronze.bronze_orders
GROUP BY order_status
ORDER BY count DESC;

-- Date range
SELECT 
    MIN(order_purchase_timestamp) as earliest_order,
    MAX(order_purchase_timestamp) as latest_order,
    DATEDIFF(MAX(order_purchase_timestamp), MIN(order_purchase_timestamp)) as days_span
FROM workspace.retail_bronze.bronze_orders;

-- Check for duplicates
SELECT 
    order_id,
    COUNT(*) as duplicate_count
FROM workspace.retail_bronze.bronze_orders
GROUP BY order_id
HAVING COUNT(*) > 1;
```
