```sql
-- ====================================================================
-- PRODUCTS TABLE PROFILING
-- ====================================================================

-- Basic counts
SELECT 
    COUNT(*) as total_products,
    COUNT(DISTINCT product_id) as unique_products,
    COUNT(DISTINCT product_category_name) as unique_categories
FROM workspace.retail_bronze.bronze_products;

-- Category distribution (top 10)
SELECT 
    product_category_name,
    COUNT(*) as product_count
FROM workspace.retail_bronze.bronze_products
WHERE product_category_name IS NOT NULL
GROUP BY product_category_name
ORDER BY product_count DESC
LIMIT 10;

-- Null analysis
SELECT 
    COUNT(*) - COUNT(product_category_name) as null_category,
    COUNT(*) - COUNT(product_weight_g) as null_weight,
    COUNT(*) - COUNT(product_length_cm) as null_length,
    COUNT(*) - COUNT(product_height_cm) as null_height,
    COUNT(*) - COUNT(product_width_cm) as null_width
FROM workspace.retail_bronze.bronze_products;

-- Product dimension statistics
SELECT 
    AVG(product_weight_g) as avg_weight,
    AVG(product_length_cm) as avg_length,
    AVG(product_height_cm) as avg_height,
    AVG(product_width_cm) as avg_width
FROM workspace.retail_bronze.bronze_products;

-- Check for invalid dimensions
SELECT COUNT(*) as invalid_dimensions
FROM workspace.retail_bronze.bronze_products
WHERE product_weight_g <= 0 
   OR product_length_cm <= 0 
   OR product_height_cm <= 0 
   OR product_width_cm <= 0;
```
