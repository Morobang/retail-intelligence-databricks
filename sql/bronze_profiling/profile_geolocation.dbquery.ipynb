```sql
-- ====================================================================
-- GEOLOCATION TABLE PROFILING
-- ====================================================================

-- Basic counts
SELECT 
    COUNT(*) as total_records,
    COUNT(DISTINCT geolocation_zip_code_prefix) as unique_zip_codes,
    COUNT(DISTINCT geolocation_city) as unique_cities,
    COUNT(DISTINCT geolocation_state) as unique_states
FROM workspace.retail_bronze.bronze_geolocation;

-- Coordinate ranges
SELECT 
    MIN(geolocation_lat) as min_latitude,
    MAX(geolocation_lat) as max_latitude,
    MIN(geolocation_lng) as min_longitude,
    MAX(geolocation_lng) as max_longitude
FROM workspace.retail_bronze.bronze_geolocation;

-- Null analysis
SELECT 
    COUNT(*) - COUNT(geolocation_zip_code_prefix) as null_zip,
    COUNT(*) - COUNT(geolocation_lat) as null_lat,
    COUNT(*) - COUNT(geolocation_lng) as null_lng,
    COUNT(*) - COUNT(geolocation_city) as null_city,
    COUNT(*) - COUNT(geolocation_state) as null_state
FROM workspace.retail_bronze.bronze_geolocation;

-- Check for invalid coordinates
SELECT COUNT(*) as invalid_coordinates
FROM workspace.retail_bronze.bronze_geolocation
WHERE geolocation_lat NOT BETWEEN -90 AND 90
   OR geolocation_lng NOT BETWEEN -180 AND 180;
```
