```sql
-- ====================================================================
-- SELLERS TABLE PROFILING
-- ====================================================================

-- Basic counts
SELECT 
    COUNT(*) as total_sellers,
    COUNT(DISTINCT seller_id) as unique_sellers,
    COUNT(DISTINCT seller_city) as unique_cities,
    COUNT(DISTINCT seller_state) as unique_states
FROM workspace.retail_bronze.bronze_sellers;

-- Seller distribution by state
SELECT 
    seller_state,
    COUNT(*) as seller_count,
    ROUND(COUNT(*) * 100.0 / (SELECT COUNT(*) FROM workspace.retail_bronze.bronze_sellers), 2) as percentage
FROM workspace.retail_bronze.bronze_sellers
GROUP BY seller_state
ORDER BY seller_count DESC;

-- Null analysis
SELECT 
    COUNT(*) - COUNT(seller_id) as null_seller_id,
    COUNT(*) - COUNT(seller_zip_code_prefix) as null_zip,
    COUNT(*) - COUNT(seller_city) as null_city,
    COUNT(*) - COUNT(seller_state) as null_state
FROM workspace.retail_bronze.bronze_sellers;

-- Check for duplicates
SELECT 
    seller_id,
    COUNT(*) as duplicate_count
FROM workspace.retail_bronze.bronze_sellers
GROUP BY seller_id
HAVING COUNT(*) > 1;
```
