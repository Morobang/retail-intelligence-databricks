```sql
-- ====================================================================
-- CATEGORY TRANSLATION TABLE PROFILING
-- ====================================================================

-- Basic counts
SELECT 
    COUNT(*) as total_records,
    COUNT(DISTINCT product_category_name) as unique_portuguese_categories,
    COUNT(DISTINCT product_category_name_english) as unique_english_categories
FROM workspace.retail_bronze.bronze_category_translation;

-- Show all category translations
SELECT 
    product_category_name,
    product_category_name_english
FROM workspace.retail_bronze.bronze_category_translation
ORDER BY product_category_name;

-- Null analysis
SELECT 
    COUNT(*) - COUNT(product_category_name) as null_portuguese,
    COUNT(*) - COUNT(product_category_name_english) as null_english
FROM workspace.retail_bronze.bronze_category_translation;

-- Check for duplicates
SELECT 
    product_category_name,
    COUNT(*) as duplicate_count
FROM workspace.retail_bronze.bronze_category_translation
GROUP BY product_category_name
HAVING COUNT(*) > 1;
```
