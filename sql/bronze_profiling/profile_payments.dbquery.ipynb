```sql
-- ====================================================================
-- PAYMENTS TABLE PROFILING
-- ====================================================================

-- Basic counts
SELECT 
    COUNT(*) as total_payments,
    COUNT(DISTINCT order_id) as unique_orders
FROM workspace.retail_bronze.bronze_order_payments;

-- Payment type distribution
SELECT 
    payment_type,
    COUNT(*) as count,
    ROUND(COUNT(*) * 100.0 / (SELECT COUNT(*) FROM workspace.retail_bronze.bronze_order_payments), 2) as percentage,
    AVG(payment_value) as avg_payment_value
FROM workspace.retail_bronze.bronze_order_payments
GROUP BY payment_type
ORDER BY count DESC;

-- Payment installments analysis
SELECT 
    payment_installments,
    COUNT(*) as count,
    AVG(payment_value) as avg_value
FROM workspace.retail_bronze.bronze_order_payments
GROUP BY payment_installments
ORDER BY payment_installments;

-- Payment value statistics
SELECT 
    MIN(payment_value) as min_payment,
    MAX(payment_value) as max_payment,
    AVG(payment_value) as avg_payment,
    PERCENTILE(payment_value, 0.5) as median_payment
FROM workspace.retail_bronze.bronze_order_payments;

-- Check for invalid payment values
SELECT COUNT(*) as invalid_payment_count
FROM workspace.retail_bronze.bronze_order_payments
WHERE payment_value <= 0;
```
