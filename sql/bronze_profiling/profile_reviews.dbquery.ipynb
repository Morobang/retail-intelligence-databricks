```sql
-- ====================================================================
-- REVIEWS TABLE PROFILING
-- ====================================================================

-- Basic counts
SELECT 
    COUNT(*) as total_reviews,
    COUNT(DISTINCT review_id) as unique_reviews,
    COUNT(DISTINCT order_id) as unique_orders_reviewed
FROM workspace.retail_bronze.bronze_order_reviews;

-- Review score distribution
SELECT 
    review_score,
    COUNT(*) as count,
    ROUND(COUNT(*) * 100.0 / (SELECT COUNT(*) FROM workspace.retail_bronze.bronze_order_reviews), 2) as percentage
FROM workspace.retail_bronze.bronze_order_reviews
GROUP BY review_score
ORDER BY review_score;

-- Average review score
SELECT 
    AVG(review_score) as avg_review_score,
    MIN(review_score) as min_score,
    MAX(review_score) as max_score
FROM workspace.retail_bronze.bronze_order_reviews;

-- Null analysis
SELECT 
    COUNT(*) - COUNT("review_comment_title") as null_comment_title,
    COUNT(*) - COUNT("review_comment_message") as null_comment_message,
    COUNT(*) - COUNT("review_creation_date") as null_creation_date,
    COUNT(*) - COUNT("review_answer_timestamp") as null_answer_timestamp
FROM workspace.retail_bronze.bronze_order_reviews;

-- Check for invalid review scores
SELECT COUNT(*) as invalid_scores
FROM workspace.retail_bronze.bronze_order_reviews
WHERE review_score NOT BETWEEN 1 AND 5;
```
